# Q1. Display First Rows

## Problem Statement
Create the following DataFrame:

| Name | Age |
|------|-----|
| Aman | 21 |
| Rahul | 22 |
| Neha | 20 |
| Priya | 23 |
| Ravi | 24 |

Display the first 3 rows using `head()`.

## Sample Output
```python
First 3 rows of DataFrame
```


````python
# Write your code here
import pandas as pd
df = pd.DataFrame({'Name':['Aman','Rahul','Neha','Priya','Ravi'],'Age':[21,22,20,23,24]})
df.head(3)
````
