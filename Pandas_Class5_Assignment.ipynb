## Q1. Sort Age in Ascending Order

Input:
Age = [25, 18, 30, 22]

Output:
18
22
25
30

```python
# Write your code here
import pandas as pd
d=pd.DataFrame({"Age":[25, 18, 30, 22],

                })
d.sort_values(by="Age")
```

## Q2. Sort Marks in Descending Order

```python
# Write your code hereimport pandas as pd
d=pd.DataFrame({
                 "Marks" : [70, 95, 80, 60]
                 })
print(d.sort_values(by="Marks",ascending=False))
```

## Q9. Find Percentage Frequency

## Q14. Student Statistics

Input:

## Q15. Mini Practice

Create a DataFrame with:
Name, Age, Score, Height

Tasks:
1. Sort by Score (Descending)
